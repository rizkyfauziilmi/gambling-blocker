```python
from urllib.parse import unquote
import re
import pandas as pd
from IPython.display import display, HTML
```

## Pra-pemrosesan Teks  
Demo fungsi `clean_url()` pada URL gambling: `https://Dunia777-V3.com/Slots/Pg%20Soft`
*Huruf kapital sengaja ditambahkan untuk mendemonstrasikan langkah lowercasing.*

```python
def clean_url(url_string):
    """Fungsi preprocessing teks sesuai pipeline classification."""
    url = unquote(url_string).lower()
    url = re.sub(r"https?:\/\/", "", url)
    url = re.sub(r"[-_/]", " ", url)
    url = re.sub(r"[^a-zA-Z0-9\s]", " ", url)
    url = re.sub(r"(\d)([a-z])", r"\1 \2", url)
    url = re.sub(r"([a-z])(\d)", r"\1 \2", url)
    url = re.sub(r"\s+", " ", url)
    url = url.strip()
    return url

def step_table(url_string):
    """Cetak tabel 7 langkah preprocessing."""
    rows = []
    
    s = url_string
    rows.append(("0. URL Mentah", "—", s))
    
    s = unquote(s)
    rows.append(("1. URL Decoding", "`%XX` \u2192 karakter asli", s))
    
    s = s.lower()
    rows.append(("2. Lowercasing", "A-Z \u2192 a-z", s))
    
    s = re.sub(r"https?:\/\/", "", s)
    rows.append(("3. Hapus Protokol", "`https://` dihapus", s))
    
    s = re.sub(r"[-_/]", " ", s)
    rows.append(("4. Replace Separator", "`- _ /` \u2192 spasi", s))
    
    s = re.sub(r"[^a-zA-Z0-9\s]", " ", s)
    rows.append(("5. Hapus Non-AN", "hapus `^a-zA-Z0-9`", s))
    
    s = re.sub(r"(\d)([a-z])", r"\1 \2", s)
    s = re.sub(r"([a-z])(\d)", r"\1 \2", s)
    rows.append(("6. Pisah Alfanumerik", "pisah huruf-digit", s))
    
    s = re.sub(r"\s+", " ", s).strip()
    rows.append(("7. Normalisasi Spasi", "trim + spasi tunggal", s))
    
    df = pd.DataFrame(rows, columns=["Langkah", "Operasi", "Hasil"])
    df.set_index("Langkah", inplace=True)
    display(df)
    
    return s

url = "https://Dunia777-V3.com/Slots/Pg%20Soft"
print(f"Input: {url}\n")
cleaned = step_table(url)
print(f"\nOutput clean_url(): {cleaned}")
```
