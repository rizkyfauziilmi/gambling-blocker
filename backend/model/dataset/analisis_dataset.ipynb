# Analisis Dataset Crawl

Menganalisis `dataset_crawl_070526_with_images_path_with_features.csv`

```python
import pandas as pd
from pathlib import Path

CSV = Path('dataset_crawl_070526_with_images_path_with_features.csv')
df = pd.read_csv(CSV)
```

```python
# Normalisasi domain: strip www. agar konsisten dengan links.py
df['Domain_Normalized'] = df['Domain'].str.lower().str.replace(r'^www\.', '', regex=True)

total = len(df)
unique_domains_norm = df['Domain_Normalized'].nunique()
gambling = (df['Tag'] == 'gambling').sum()
non_gambling = (df['Tag'] == 'non-gambling').sum()
urls_per_domain = df.groupby('Domain_Normalized').size()
max_url = urls_per_domain.max()

print('='*60)
print('DATA TEKS (seluruh dataset)')
print('='*60)
print(f'Total alamat situs        : {total:,}')
print(f'Domain unik (strip www)   : {unique_domains_norm:,}')
print(f'Situs judi                : {gambling:,}')
print(f'Situs non-judi            : {non_gambling:,}')
print(f'Maksimal URL/domain       : {max_url}')
```

```python
has_screenshot = df['Screenshot_Path'].notna() & (df['Screenshot_Path'] != '')
img = df[has_screenshot].copy()

img_total = len(img)
img_domains_norm = img['Domain_Normalized'].nunique()
img_gambling = (img['Tag'] == 'gambling').sum()
img_non_gambling = (img['Tag'] == 'non-gambling').sum()

print('='*60)
print('DATA GAMBAR (Screenshot_Path terisi)')
print('='*60)
print(f'Total screenshot         : {img_total:,}')
print(f'Domain unik (strip www)  : {img_domains_norm:,}')
print(f'Situs judi               : {img_gambling:,}')
print(f'Situs non-judi           : {img_non_gambling:,}')
```
